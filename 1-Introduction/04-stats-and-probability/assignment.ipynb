### Task 5: Test the hypothesis that the degree of diabetes progression is different between men and women

```python
from scipy.stats import ttest_ind
```

```python
t_stat, p_val = ttest_ind(df.loc[df['SEX'] == 1, ['Y']], 
                          df.loc[df['SEX'] == 2, ['Y']], 
                          equal_var=False)
print(f"T_value = {t_stat[0]:2f}\nP_value: {p_val[0]}")
```

Conclusión:

La media de ambos grupos es muy cercana, la media 2 es mayor que la 1
Ya que P_value > 0.05, no hay evidencia suficiente para afirmar que existe una diferencia real.

Conclusion:

The mean of both groups is very close, mean 2 is higher than mean 1.
However, since the p_value is greater than 0.05, there is insufficient evidence to conclude thet a real difference exist
